```python
import numpy as np
from sklearn.neural_network import MLPClassifier
```

```python
#데이터 정의
X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = [0, 1, 1, 0]  # XOR 문제
```

```python
# 다층 퍼셉트론 모델 정의
classifier = MLPClassifier(hidden_layer_sizes=(2,), activation='logistic',learning_rate_init=0.1)
```

**Method1. fit함수 이용**

```python
#model 학습
# fit 함수 이용해서 학습 -> 가중치 초기화, 손실함수 계산, 가중치 업데이트가 모두 포함되어 있음
classifier.fit(X, y)
w= classifier.coefs_  # 가중치 확인
w_0= classifier.intercepts_  # 편향 확인
print(w)
print(w_0)
```

```python
# 모델 성능 평가
# predict 함수 이용해서 예측
y_pred = classifier.predict(X)
print(y_pred)
```

```python
# 데이터 정의
# XOR데이터 정의, 두 입력값이 같으면 0, 다르면 1
X = np.array([0,0,1,1,0,1,0,1]).reshape(2,4)
y=np.array([0,1,1,0]).reshape(1,4)
print(X)
```

**Method2. scikit-learn 이용**

```python
# 다층퍼셉트론 모델 정의
# scikit-learn 라이브러리 이용해서 구현되어 있는 다층퍼셉트론 모델 불러옴 -> 가중치 초기화, 가중합 함수, 활성화 함수, 손실함수,
# 가중치 업데이트 모두 포함되어 있음

np.random.seed(59)
W_0 = np.random.rand(2,2)  # 입력층과 은닉층 사이의 가중치
W_1 = np.random.rand(2,1)  # 은닉층과 출력
B_0 = np.random.random((1,2))  # 은닉층 편향
B_1 = np.random.random((1,1))  # 출력층 편향
lr = 0.1  # 학습률

def weighted_sum(X, W, B):
    return np.dot(W.T, X) + B.T
def sigmoid(x):
    return 1/(1+np.exp(-x))
def BCE_loss(y, y_hat):
    loss=np.mean(y*(np.log(y_hat)) + (1-y)*(np.log(1-y_hat)))
    return -loss
def gradient_update(X, H, y, y_hat):
    global W_0,B_0, W_1,B_1, lr
    dW_1=np.dot(H, (y_hat-y).T)
    dB_1=np.sum(y_hat-y, axis=1, keepdims=True).T
    dH=np.dot(W_1, (y_hat-y))

    dZ_1 = dH*(H*(1-H))
    dW_0=np.dot(X, dZ_1.T)
    dB_0=np.sum(dZ_1, axis=1, keepdims=True).T

    W_0=W_0-lr*dW_0
    W_1=W_1-lr*dW_1
    B_0=B_0-lr*dB_0
    B_1=B_1-lr*dB_1
```

> 이때 실수한 점. B_0는 (1,2)행렬, 그래서 가중합 계산할 때 편향은 Transpose 시켜야 행렬 계산이 가능해짐.

```python
# 모델 학습
# 위 다섯개 함수 모두 이용해서 모델 학습

for epoch in range(1000):
    Z_1 = weighted_sum(X, W_0, B_0)
    H = sigmoid(Z_1)
    Z_2 = weighted_sum(H, W_1, B_1)
    y_hat = sigmoid(Z_2)

    loss = BCE_loss(y, y_hat)
    gradient_update(X, H, y, y_hat)

print(W_0)
print(W_1)
print(B_0)
print(B_1)
```

```python
# 모델 성능 평가
def predict(X):
    global W_0, W_1, B_0, B_1
    Z_1 = weighted_sum(X, W_0, B_0)
    H = sigmoid(Z_1)
    Z_2 = weighted_sum(H, W_1, B_1)
    y_hat = sigmoid(Z_2).reshape(4)

    pred_class =[]
    pred_class = [1 if i > 0.5 else 0 for i in y_hat]
    return np.array(pred_class)

y_pred = predict(X)
print(y_pred)
```
